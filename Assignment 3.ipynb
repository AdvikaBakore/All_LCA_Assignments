```python
#To check whether or not the triangle is is a right angled triangle using function:
"""Aim:Write a python program which accepts the length of three sides of a triangle as inputs.
       The program should indicate whether or not the triangle is a right angled triangle using function."""
```

```python
def check_right_angled_triangle(a,b,c):
    largest=max(a,b,c)

    if largest==a:
        if b**2+c**2==a**2:
            return True
    elif largest==b:
        if a**2+c**2==b**2:
            return True
    else:
        if largest==c:
            if a**2+b**2==c**2:
                return True
    return False

a=float(input("Enter length of first side:"))
b=float(input("Enter length of second side:"))
c=float(input("Enter length of third side:"))

if check_right_angled_triangle(a,b,c):
    print("It is a right angled triangle")
else:
    print("It is not right angled triangle")
```
