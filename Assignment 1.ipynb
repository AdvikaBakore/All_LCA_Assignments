```python
# Create list:

student_list=["Advika","Bhoomi","Saee","Sejal"]
print("\nStudent List:",student_list)

#Create Tuple:

student_tuple=("Sejal","Aadya","Advika","Anushka")
print("\nStudent Tuple:",student_tuple)

#Create Dictionary:

student_dict={10:"Advika",11:"Aashna",12:"Aahana",13:"Aarushi"}
print("\nStudent Dictionary:",student_dict)
```

```python
#Adding an element in a dictionary:
student_dict[14]="Siya"
print("\nAfter Adding:",student_dict)

#Updating an element in a dictionary:
student_dict[10]="Vedha"
print("\nAfter Updating:",student_dict)

#Deleting an element from the dictionary:
del student_dict[10]
print("\nAfter Deleting:",student_dict)
```
